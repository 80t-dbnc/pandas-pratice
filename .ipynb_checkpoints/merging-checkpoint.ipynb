```python
import numpy as np
import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
# inner mein jo uske andar th bss vahi add hua
#bss drop hii lgta hai row wise hota hai
```

```python
pd.merge(employees,salaries,on = "employee_id",how = "inner")
```

```python
#outwr mein jo nahi hai vo sbb bhi merge ho gyi
```

```python
pd.merge(employees,salaries,on = "employee_id",how = "outer")
```

```python
#so merge to employee id ke basis pe hogi but usme bhi jiske pass right wala in btwn enployee and salary 
#yani salary wali data frame add hogi ye to proper bakchodi hai bhosdi ke 
#cod eke naam pe chutiya kata ja rha hai
```

```python
pd.merge(employees,salaries,on = "employee_id",how = "right")
```

```python
# left yani left wala in btwn employye and salary uspe run pe hogi
# so employee id ke sath jo employee wale dataframe ke  honge vo sbb add honge
# kuch khas nhi hai bhosdi ke bss bakchodi hai isme
```

```python
pd.merge(employees,salaries,on = "employee_id",how = "left")
```

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df1.join(df2,how = "outer")
```

```python
df2.join(df1)
```
